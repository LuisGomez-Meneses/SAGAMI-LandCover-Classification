```python
# ==============================================================================
# SCRIPT 08
# RANDOM FOREST
# PARTE 1
# CARGA DE DATOS
# ==============================================================================

import joblib
import numpy as np
import pandas as pd

from pathlib import Path

from sklearn.ensemble import RandomForestClassifier

# ==============================================================================
# CONFIGURACIÓN
# ==============================================================================

#TARGET = "clase_n1"

TARGET = "clase_n2"
# TARGET = "clase_n3"

RANDOM_STATE = 42

N_ESTIMATORS = 500

# ==============================================================================
# RUTAS
# ==============================================================================

ML = Path("/content/drive/MyDrive/ML")

MODELS = Path("/content/Modelos")

MODELS.mkdir(exist_ok=True)

# ==============================================================================
# CARGAR DATOS
# ==============================================================================

print("="*80)
print("SCRIPT 08")
print("RANDOM FOREST")
print("="*80)

print(f"\nNivel seleccionado : {TARGET}")

print("\nCargando datos...")

X_train = pd.read_parquet(
    ML / f"X_train_{TARGET}.parquet"
)

X_test = pd.read_parquet(
    ML / f"X_test_{TARGET}.parquet"
)

y_train = np.load(
    ML / f"y_train_{TARGET}.npy"
)

y_test = np.load(
    ML / f"y_test_{TARGET}.npy"
)

encoder = joblib.load(
    ML / f"LabelEncoder_{TARGET}.joblib"
)

class_weights = joblib.load(
    ML / f"ClassWeights_{TARGET}.joblib"
)

# ==============================================================================
# INFORMACIÓN
# ==============================================================================

print("\n" + "="*80)
print("RESUMEN")
print("="*80)

print(f"X_train : {X_train.shape}")

print(f"X_test  : {X_test.shape}")

print(f"y_train : {y_train.shape}")

print(f"y_test  : {y_test.shape}")

print(f"\nNúmero de clases : {len(encoder.classes_)}")

print("\nClases:")

for i, c in enumerate(encoder.classes_):

    print(f"{i:2d} -> {c}")

print("\nVariables predictoras:")

for variable in X_train.columns:

    print("-", variable)

# ==============================================================================
# CONFIGURACIÓN DEL RANDOM FOREST
# ==============================================================================

rf = RandomForestClassifier(

    n_estimators=N_ESTIMATORS,

    criterion="gini",

    max_features="sqrt",

    bootstrap=True,

    oob_score=True,

    random_state=RANDOM_STATE,

    n_jobs=-1,

    class_weight=class_weights

)

print("\n" + "="*80)
print("MODELO CONFIGURADO")
print("="*80)

print(rf)
```

```python
# ==============================================================================
# SCRIPT 08
# PARTE 2
# ENTRENAMIENTO
# ==============================================================================

import time
import joblib

print("\n" + "="*80)
print("ENTRENANDO RANDOM FOREST")
print("="*80)

inicio = time.time()

rf.fit(

    X_train,

    y_train

)

fin = time.time()

tiempo = fin - inicio

print("\nEntrenamiento finalizado.")

print(f"\nTiempo de entrenamiento : {tiempo:.2f} segundos")

print(f"OOB Score : {rf.oob_score_:.4f}")

print(f"Número de árboles : {len(rf.estimators_)}")

# ==============================================================================
# PROFUNDIDAD DE LOS ÁRBOLES
# ==============================================================================

profundidades = [

    arbol.tree_.max_depth

    for arbol in rf.estimators_

]

print(f"\nProfundidad promedio : {np.mean(profundidades):.1f}")

print(f"Profundidad máxima   : {np.max(profundidades)}")

print(f"Profundidad mínima   : {np.min(profundidades)}")

# ==============================================================================
# GUARDAR MODELO
# ==============================================================================

modelo = MODELS / f"RandomForest_{TARGET}.joblib"

joblib.dump(

    rf,

    modelo

)

print("\nModelo guardado:")

print(modelo)

print("\n" + "="*80)
print("MODELO LISTO PARA PREDICCIÓN")
print("="*80)
```
