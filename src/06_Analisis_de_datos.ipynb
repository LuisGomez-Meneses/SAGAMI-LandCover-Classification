```python
# ==============================================================================
# SCRIPT 06
# ANÁLISIS DEL DATASET
# PARTE 1
# ==============================================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
from pathlib import Path

# ==============================================================================
# RUTAS
# ==============================================================================

DATASET = Path("/content/Dataset_SAGAMI.csv")

OUTPUT = Path("/content/Reportes")

OUTPUT.mkdir(exist_ok=True)

# ==============================================================================
# CARGAR DATASET
# ==============================================================================

print("="*80)
print("SCRIPT 06")
print("ANÁLISIS DEL DATASET")
print("="*80)

print("\nCargando dataset...")

df = pd.read_csv(DATASET)

print("OK")

# ==============================================================================
# INFORMACIÓN GENERAL
# ==============================================================================

print("\n" + "="*80)
print("INFORMACIÓN GENERAL")
print("="*80)

print(f"Número de muestras : {len(df):,}")

print(f"Número de variables : {df.shape[1]}")

print("\nVariables:")

for c in df.columns:
    print(" -", c)

# ==============================================================================
# TIPOS DE DATOS
# ==============================================================================

print("\n" + "="*80)
print("TIPOS DE DATOS")
print("="*80)

print(df.dtypes)

# ==============================================================================
# VALORES NULOS
# ==============================================================================

print("\n" + "="*80)
print("VALORES NULOS")
print("="*80)

nulls = df.isnull().sum()

print(nulls)

# ==============================================================================
# VALORES INFINITOS
# ==============================================================================

print("\n" + "="*80)
print("VALORES INFINITOS")
print("="*80)

numeric = df.select_dtypes(include=np.number)

inf = np.isinf(numeric).sum()

print(inf)

# ==============================================================================
# MEMORIA
# ==============================================================================

print("\n" + "="*80)
print("USO DE MEMORIA")
print("="*80)

memory = df.memory_usage(deep=True).sum() / (1024**2)

print(f"{memory:.2f} MB")

# ==============================================================================
# VARIABLES ESPECTRALES
# ==============================================================================

spectral = [

    "B2",
    "B3",
    "B4",
    "B5",
    "B6",
    "B7",
    "B8",
    "B8A",
    "B11",
    "B12",
    "VV",
    "VH"

]

print("\nVariables predictoras:")

print(spectral)

# ==============================================================================
# RESUMEN
# ==============================================================================

print("\n" + "="*80)
print("RESUMEN")
print("="*80)

print(f"Dataset : {DATASET.name}")

print(f"Muestras : {len(df):,}")

print(f"Variables predictoras : {len(spectral)}")

print(f"Variables totales : {df.shape[1]}")

print("\nDataset listo para el análisis estadístico.")
```

```python
# ==============================================================================
# SCRIPT 06
# PARTE 2
# ESTADÍSTICA DESCRIPTIVA
# ==============================================================================

print("\n" + "="*80)
print("ESTADÍSTICA DESCRIPTIVA")
print("="*80)

# ------------------------------------------------------------------
# Variables predictoras
# ------------------------------------------------------------------

spectral = [
    "B2","B3","B4","B5","B6",
    "B7","B8","B8A","B11","B12",
    "VV","VH"
]

estadisticas = df[spectral].describe().T

estadisticas["CV"] = (
    estadisticas["std"] / estadisticas["mean"]
)

print(estadisticas)

estadisticas.to_csv(
    OUTPUT/"Resumen_estadistico.csv"
)

# ==============================================================================
# DISTRIBUCIÓN DE CLASES
# ==============================================================================

print("\n" + "="*80)
print("DISTRIBUCIÓN DE CLASES")
print("="*80)

for nivel in ["clase_n1","clase_n2","clase_n3"]:

    print("\n" + "-"*60)
    print(nivel)
    print("-"*60)

    tabla = (
        df[nivel]
        .value_counts()
        .rename_axis(nivel)
        .reset_index(name="pixeles")
    )

    tabla["porcentaje"] = (
        tabla["pixeles"] /
        len(df) * 100
    ).round(3)

    print(tabla)

    tabla.to_csv(
        OUTPUT/f"{nivel}_distribucion.csv",
        index=False
    )

# ==============================================================================
# CLASES PEQUEÑAS
# ==============================================================================

print("\n" + "="*80)
print("CLASES CON POCAS MUESTRAS")
print("="*80)

conteo = df["clase_n3"].value_counts()

print("\nMenos de 100 píxeles")

print(conteo[conteo < 100])

print("\nMenos de 50 píxeles")

print(conteo[conteo < 50])

print("\nMenos de 20 píxeles")

print(conteo[conteo < 20])

# ==============================================================================
# RESUMEN
# ==============================================================================

print("\n" + "="*80)
print("RESUMEN")
print("="*80)

print(f"Clases Nivel 1 : {df['clase_n1'].nunique()}")

print(f"Clases Nivel 2 : {df['clase_n2'].nunique()}")

print(f"Clases Nivel 3 : {df['clase_n3'].nunique()}")

print("\nArchivo generado:")

print(OUTPUT/"Resumen_estadistico.csv")
```
