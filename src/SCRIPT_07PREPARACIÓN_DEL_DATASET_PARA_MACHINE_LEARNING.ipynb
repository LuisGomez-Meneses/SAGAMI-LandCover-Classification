```python
# ==============================================================================
# SCRIPT 07
# PREPARACIÓN DEL DATASET PARA MACHINE LEARNING
# PARTE 1
# ==============================================================================

import joblib
import numpy as np
import pandas as pd

from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.utils.class_weight import compute_class_weight

# ==============================================================================
# CONFIGURACIÓN
# ==============================================================================

TARGET = "clase_n3"

# TARGET = "clase_n2"
# TARGET = "clase_n3"

TEST_SIZE = 0.30

RANDOM_STATE = 42

# ==============================================================================
# RUTAS
# ==============================================================================

DATASET = Path("/content/Dataset_SAGAMI.csv")

OUTPUT = Path("/content/ML")

OUTPUT.mkdir(exist_ok=True)

# ==============================================================================
# CARGAR DATASET
# ==============================================================================

print("="*80)
print("SCRIPT 07")
print("PREPARACIÓN PARA MACHINE LEARNING")
print("="*80)

df = pd.read_csv(DATASET)

print(f"\nDataset cargado : {len(df):,} muestras")

# ==============================================================================
# VARIABLES PREDICTORAS
# ==============================================================================

FEATURES = [

    "B2",
    "B3",
    "B4",
    "B5",
    "B6",
    "B7",
    "B8",
    "B8A",
    "B11",
    "B12",

    "VV",
    "VH"

]

print("\nNivel de clasificación :")

print(TARGET)

print("\nVariables predictoras :")

print(FEATURES)

# ==============================================================================
# MATRICES
# ==============================================================================

X = df[FEATURES].copy()

y = df[TARGET].copy()

print("\nDimensiones")

print("X :", X.shape)

print("y :", y.shape)

# ==============================================================================
# LABEL ENCODER
# ==============================================================================

encoder = LabelEncoder()

y_encoded = encoder.fit_transform(y)

joblib.dump(

    encoder,

    OUTPUT / f"LabelEncoder_{TARGET}.joblib"

)

print("\nClases encontradas\n")

for i, c in enumerate(encoder.classes_):

    print(f"{i:2d} -> {c}")

# ==============================================================================
# PESOS DE CLASE
# ==============================================================================

weights = compute_class_weight(

    class_weight="balanced",

    classes=np.unique(y_encoded),

    y=y_encoded

)

class_weights = {

    i: float(w)

    for i, w in enumerate(weights)

}

print("\nPesos de clase\n")

for k, v in class_weights.items():

    print(f"{encoder.classes_[k]:25s} : {v:.3f}")

# ==============================================================================
# TRAIN / TEST
# ==============================================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y_encoded,

    test_size=TEST_SIZE,

    random_state=RANDOM_STATE,

    stratify=y_encoded

)

print("\n"+"="*80)

print("DIVISIÓN DEL DATASET")

print("="*80)

print(f"Train : {len(X_train):,}")

print(f"Test  : {len(X_test):,}")

print("\nDistribución TRAIN")

print(pd.Series(y_train).value_counts().sort_index())

print("\nDistribución TEST")

print(pd.Series(y_test).value_counts().sort_index())

# ==============================================================================
# GUARDAR
# ==============================================================================

X_train.to_parquet(

    OUTPUT / f"X_train_{TARGET}.parquet",

    index=False

)

X_test.to_parquet(

    OUTPUT / f"X_test_{TARGET}.parquet",

    index=False

)

np.save(

    OUTPUT / f"y_train_{TARGET}.npy",

    y_train

)

np.save(

    OUTPUT / f"y_test_{TARGET}.npy",

    y_test

)

joblib.dump(

    class_weights,

    OUTPUT / f"ClassWeights_{TARGET}.joblib"

)

print("\nArchivos generados\n")

print(f"✔ X_train_{TARGET}.parquet")

print(f"✔ X_test_{TARGET}.parquet")

print(f"✔ y_train_{TARGET}.npy")

print(f"✔ y_test_{TARGET}.npy")

print(f"✔ LabelEncoder_{TARGET}.joblib")

print(f"✔ ClassWeights_{TARGET}.joblib")
```
